```python
words=open("data/indian_cities.txt").read().splitlines()
words=[w for w in words if len(w)<20]

words = [w.strip().lower() for w in words]
words=[w for w in words if "(" not in w and ")" not in w]
len(words)
```

```python
chars=sorted(list(set("".join(words))))
stoi={s:i+1 for i,s in enumerate(chars)}
stoi["."]=0
itos={i:s for s,i in stoi.items()}
stoi
```

```python
import torch
import torch.nn.functional as f
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
#building the dataset
block_size = 3
X,Y=[],[]

for w in words:
  # print(w)
  context=[0]*block_size
  for ch in w + '.':
    ix=stoi[ch]
    X.append(context)
    Y.append(ix)
    context=context[1:]+[ix]
X=torch.tensor(X)
Y=torch.tensor(Y)
```

```python
C=torch.randn(len(stoi),3,requires_grad=True)

# emb.shape
W1=torch.randn(9,100,requires_grad=True)
# W1.shape

b1=torch.randn(100,requires_grad=True)
# # # inp = torch.cat(
# # #     torch.unbind(emb, 1), dim=1
# # # )

# hidden_layer=torch.tanh(emb.view(len(emb),9)@W1+b1)
W2=torch.randn(100,len(stoi),requires_grad=True)
b2=torch.randn(len(stoi),requires_grad=True)
perimeters=[C,W1,b1,W2,b2]

# logits=hidden_layer@W2+b2

# loss=f.cross_entropy(logits,Y)

# loss





```

```python
#training the neural network
for epoc in range(10000):
    emb=C[X]
    # forward pass
    hidden_layer=torch.tanh(emb.view(len(emb),9)@W1+b1)
    logits=hidden_layer@W2+b2
    loss=f.cross_entropy(logits,Y)
    

    for p in perimeters:
        p.grad=None
    

    #backward pass
    loss.backward()
    for p in perimeters:
        p.data-= 0.1*p.grad
    if (epoc+1)%1000==0 :
        print(loss)
    
    
    
```

```python
def train(lr,ep):
    for epoc in range(ep):
      emb=C[X]
    # forward pass
      hidden_layer=torch.tanh(emb.view(len(emb),9)@W1+b1)
      logits=hidden_layer@W2+b2
      loss=f.cross_entropy(logits,Y)
    

      for p in perimeters:
        p.grad=None
    

    #backward pass
      loss.backward()
      for p in perimeters:
        p.data-= lr*p.grad
      if (epoc+1)%1000==0 :
        print(loss)
#learning rate decay

train(0.5,3000)
train(0.25,3000)
```

```python
# sampling from the model
import torch.nn.functional as f

for _ in range(10):

    context = [0, 0, 0]
    out = []

    while True:

        # Get embeddings for current context
        emb = C[torch.tensor(context)]

        # Forward pass
        hidden_layer = torch.tanh(emb.view(1, 9) @ W1 + b1)
        logits = hidden_layer @ W2 + b2

        # Convert logits into probabilities
        probs = f.softmax(logits, dim=1)

       
        ix = torch.multinomial(probs, num_samples=1).item()

       
        if ix == stoi['.']:
            break

        out.append(itos[ix])

        context = context[1:] + [ix]

    print(''.join(out))
```
